```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd

# Trouve la racine du projet (le dossier qui contient "data")
ROOT = Path.cwd()
while not (ROOT / "data").exists() and ROOT != ROOT.parent:
    ROOT = ROOT.parent

files = sorted((ROOT / "data" / "raw").glob("*/*.parquet"))
df = pd.concat((pd.read_parquet(f) for f in files), ignore_index=True)
df["time"] = pd.to_datetime(df["time"])
print(len(files), "fichiers,", df.shape)
```

```python
print(df.groupby("city").size())
df.groupby("city")["time"].nunique()
```

```python
print(df.groupby("city").size())
print(df.isna().sum())
df.describe().T
```

```python
RAD = "shortwave_radiation"
df["hour"] = df["time"].dt.hour
df["month"] = df["time"].dt.month
df["year"] = df["time"].dt.year
```

```python
profile = df.groupby(["city", "hour"])[RAD].mean().unstack("city")
profile.plot(figsize=(9, 4), title="Rayonnement moyen par heure (UTC)")
plt.ylabel("W/m²")
plt.show()
```
